```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def simulate(dt, nWarmUp, nSteps):
    x = 0 

    for i in range(nWarmUp):
        dW = np.random.normal(0, np.sqrt(dt))
        dx = -x * dt + dW # do the time step
        x += dx # increment our variable

    hist = []
    for i in range(nSteps):
            dW = np.random.normal(0, np.sqrt(dt))
            dx = -x * dt + dW # do the time step
            x += dx # increment our variable
            hist.append(x)

    return hist 


hist = simulate(.01, 1000, 1000_000)
plt.hist(hist, bins=124, density=True)

xx = np.linspace(-3,3)
gauss = np.exp(- xx**2) / np.sqrt(np.pi)
plt.plot(xx, gauss, '--')
plt.show()
```

```python
plt.hist(hist, bins=124, density=True, label='histogram of simulation')

xx = np.linspace(-3,3)
gauss = np.exp(- xx**2) / np.sqrt(np.pi)
plt.plot(xx, gauss, '--', label='Analytical solution: Gaussian with variance 1')
plt.xlabel("x")
plt.ylabel("p(x)")
plt.legend(loc='lower left')
plt.title("Numerical Simulation of SDE")
plt.show()
```
